# Student Employment Competitiveness & Industry Skill Gap Analyzer
## Exploratory Data Analysis (EDA)

This notebook provides a thorough exploratory analysis of the two official source datasets:
1. `College_Employment_Competitiveness.csv`
2. `IT_Job_Roles_Skills.csv`

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

### 1. Load Student Dataset

```python
df_student = pd.read_csv('../data/raw/College_Employment_Competitiveness.csv')
print('Dataset Shape:', df_student.shape)
print('Missing Values:', df_student.isnull().sum().sum())
print('Duplicate Rows:', df_student.duplicated().sum())
df_student.head()
```

### 2. Target Variable Distribution

```python
print(df_student['Employment_Competitiveness'].value_counts())
plt.figure(figsize=(6, 3.5))
df_student['Employment_Competitiveness'].value_counts().plot(kind='bar', color=['#2563EB', '#16A34A', '#DC2626'])
plt.title('Target Distribution (Employment Competitiveness)')
plt.xlabel('Category')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```

### 3. Summary Statistics of Student Features

```python
df_student.describe().T[['min', 'mean', '50%', 'max', 'std']]
```

### 4. Correlation Matrix

```python
plt.figure(figsize=(10, 8))
sns.heatmap(df_student.drop(columns=['Employment_Competitiveness']).corr(), annot=True, fmt='.2f', cmap='Blues')
plt.title('Student Feature Correlation Matrix')
plt.tight_layout()
plt.show()
```

### 5. Load IT Job Roles and Skills Dataset

```python
df_jobs = pd.read_csv('../data/raw/IT_Job_Roles_Skills.csv', encoding='cp1252')
print('Jobs Shape:', df_jobs.shape)
print('Unique Job Titles:', df_jobs['Job Title'].nunique())
df_jobs.head(5)
```

### 6. Top Industry Skills Frequency

```python
from collections import Counter
all_skills = []
for s in df_jobs['Skills'].dropna():
    parts = [p.strip() for p in str(s).replace(';', ',').split(',') if p.strip()]
    all_skills.extend(parts)

skill_counts = Counter(all_skills)
top_20_skills = pd.Series(dict(skill_counts.most_common(20)))

plt.figure(figsize=(9, 5))
top_20_skills.sort_values().plot(kind='barh', color='#2563EB')
plt.title('Top 20 Most Frequent Industry Skills')
plt.xlabel('Number of Mentions')
plt.tight_layout()
plt.show()
```
